## Zadanie 1 (1 pkt)

Napisz funkcję **bin2dec** przyjmującą string będący liczbą zapisaną w postaci binarnej a zwracającą liczbę zapisaną w postaci dziesiętnej oraz funkcję **dec2bin()** przyjmującą liczbę typu int a zwracającą string będący liczbą zapisaną w postaci binarnej, funkcja ta ma przyjmować również drugi argument 'pad', który będzie określał do jakiej długości binarna postać będzie uzupełniana zerami z przodu.

```python
def bin2dec(bin_str):
    dec = 0
    for i in range(len(bin_str)):
      if bin_str[i] == '1':
        dec += 2 ** (len(bin_str) - i - 1)

    return dec
```

```python
print(bin2dec('0')==0)
print(bin2dec('10')==2)
print(bin2dec('111')==7)
print(bin2dec('1001')==9)
print(bin2dec('0001')==1)
```

```python
def dec2bin(dec,pad):
    bin_str = ""
    while dec > 0:
      if dec % 2 == 1:
        dec -= 1
        bin_str += "1"
      else:
        bin_str += "0"
      dec /= 2

    while len(bin_str) < pad:
      bin_str += "0"

    return bin_str[::-1]
```

```python
print(dec2bin(0,1)=='0')
print(dec2bin(2,2)=='10')
print(dec2bin(7,4)=='0111')
print(dec2bin(9,4)=='1001')
print(dec2bin(1,4)=='0001')
```

Napisz funkcję **shift_left(tab,n)**, która przyjmuje tablice **tab** oraz liczbę całkowitą **n** i zwraca tablicę, która jest pierwotną tablicą z elementami przesuniętymi w lewo o **n**.

```python
def shift_left(tab,n):
    temp_tab = tab[:]
    for i in range(len(tab)):
      tab[(i - n) % len(tab)] = temp_tab[i]

    return tab
```

```python
print(shift_left([1,2,3,4],2)==[3,4,1,2])
print(shift_left([1,2,3,4],3)==[4,1,2,3])
print(shift_left([1,2,3,4,5,6],2)==[3,4,5,6,1,2])
```

Napisz funkcję **permute()**, która przyjmuje string **k** i dokonuje permutacji znaków zgodnie z tablicą **perm**, w tym równeż tzw. permutację rozszerzoną, która duplikuje niektóre elementy oraz taką, która pomija niektóre elementy, patrz przykłady w testach:

```python
def permute(k,perm):
    out_str = ""
    for p in perm:
      out_str += k[p]
    return out_str
```

```python
print(permute('abcd',[1,0,2,3])=='bacd')
print(permute('abcd',[3,2,1,0])=='dcba')
print(permute('abcd',[0,1,2,3,0])=='abcda')
print(permute('1100',[0,1,3,2,0,3])=='110010')
print(permute('1100',[0,2,1])=='101')
```

Napisz funkcję **xor()**, która przyjmuje dwa równej długości zmienne typu string **bin_str1** i **bin_str2** będące liczbami zapisanymi w postaci binarnej i dokonuje operacji alternatywy wykluczającej (1+0=0,1+1=0,0+0=0) a następnie zwraca string z wynikową liczbą w postaci binarnej.

```python
def xor(bin_str1,bin_str2):
    out_str = ""
    for i in range(len(bin_str1)):
      if bin_str1[i] == bin_str2[i]:
        out_str += "0"
      else:
        out_str += "1"

    return out_str
```

```python
print(xor('0','0')=='0')
print(xor('1','1')=='0')
print(xor('01','11')=='10')
print(xor('1101','1111')=='0010')
print(xor('11111','11111')=='00000')
```
